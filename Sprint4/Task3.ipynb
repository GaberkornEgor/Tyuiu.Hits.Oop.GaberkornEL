```
//Интерфейс для работы с книгами
interface IBook
{
    string GetInfo();
    void Read();
    void Borrow();
}

//Интерфейс для дополнительных действий с книгой
interface IBookActions
{
    void ShowShortInfo();
    void ReadBook();
}

//Интерфейс для прослушивания аудиокниг
interface IAudioBook
{
    void Listen();
}

//Интерфейс для скачивания книг
interface IDownloadable
{
    void Download();
}

//Базовый класс Book
//Реализует интерфейсы IBook и IBookActions
class Book : IBook, IBookActions
{
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }

    //4 дополнительных атрибута
    public int Pages { get; set; }
    public string Publisher { get; set; }
    public string Language { get; set; }
    public double Rating { get; set; }

    //Конструктор
    public Book(string title, string author, int year,
                int pages, string publisher,
                string language, double rating)
    {
        Title = title;
        Author = author;
        YearOfPublication = year;
        Pages = pages;
        Publisher = publisher;
        Language = language;
        Rating = rating;
    }

    //Вывод информации о книге
    public virtual string GetInfo()
    {
        return "Название: " + Title +
               ", Автор: " + Author +
               ", Год издания: " + YearOfPublication;
    }

    //Виртуальный метод чтения книги
    public virtual void Read()
    {
        Console.WriteLine("Книга " + Title + " читается.");
    }

    //Виртуальный метод выдачи книги
    public virtual void Borrow()
    {
        Console.WriteLine("Книга " + Title + " выдана на чтение.");
    }

    //Явная реализация интерфейса IBookActions
    //Эти методы доступны через ссылку на интерфейс
    void IBookActions.ShowShortInfo()
    {
        Console.WriteLine("Краткая информация: " + Title + " - " + Author);
    }

    void IBookActions.ReadBook()
    {
        Read();
    }

    //4 дополнительных метода

    //1. Вывод количества страниц
    public void ShowPages()
    {
        Console.WriteLine("Количество страниц: " + Pages);
    }

    //2. Вывод издательства
    public void ShowPublisher()
    {
        Console.WriteLine("Издательство: " + Publisher);
    }

    //3. Вывод языка
    public void ShowLanguage()
    {
        Console.WriteLine("Язык книги: " + Language);
    }

    //4. Вывод рейтинга
    public void ShowRating()
    {
        Console.WriteLine("Рейтинг книги: " + Rating);
    }

    //Сравнение года издания с другой книгой
    public void CompareYear(Book book)
    {
        if (YearOfPublication > book.YearOfPublication)
        {
            Console.WriteLine(Title + " новее, чем " + book.Title);
        }
        else if (YearOfPublication < book.YearOfPublication)
        {
            Console.WriteLine(Title + " старше, чем " + book.Title);
        }
        else
        {
            Console.WriteLine(Title + " и " + book.Title +
                              " изданы в одном году.");
        }
    }

    //Перегрузка метода CompareYear
    //Сравнение года издания с указанным годом
    public void CompareYear(int year)
    {
        if (YearOfPublication > year)
        {
            Console.WriteLine(Title + " новее книги, изданной в " + year + " году.");
        }
        else if (YearOfPublication < year)
        {
            Console.WriteLine(Title + " старше книги, изданной в " + year + " году.");
        }
        else
        {
            Console.WriteLine(Title + " издана в " + year + " году.");
        }
    }

    //Взаимодействие двух объектов Book
    public void Recommend(Book book)
    {
        Console.WriteLine("Книга " + Title +
                          " рекомендуется вместе с книгой " + book.Title);
    }

    //Перегрузка метода Read
    //Чтение определенного количества страниц
    public void Read(int pages)
    {
        Console.WriteLine("Читаем " + pages +
                          " страниц книги " + Title);
    }

    //Перегрузка метода Read
    //Чтение книги в указанном формате
    public void Read(string format)
    {
        Console.WriteLine("Книга " + Title +
                          " читается в формате " + format);
    }
}

//Простое наследование
//Textbook наследуется от Book
class Textbook : Book
{
    public string Subject { get; set; }

    //4 дополнительных атрибута
    public string Level { get; set; }
    public int ExercisesCount { get; set; }
    public bool HasAnswers { get; set; }
    public string Department { get; set; }

    public Textbook(string title, string author, int year,
                    int pages, string publisher,
                    string language, double rating,
                    string subject, string level,
                    int exercisesCount, bool hasAnswers,
                    string department)
        : base(title, author, year, pages, publisher, language, rating)
    {
        Subject = subject;
        Level = level;
        ExercisesCount = exercisesCount;
        HasAnswers = hasAnswers;
        Department = department;
    }

    //Перекрытие метода Read
    public override void Read()
    {
        Console.WriteLine("Учебник " + Title +
                          " читается по предмету: " + Subject);
    }

    //Перекрытие метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Предмет: " + Subject +
               ", Уровень: " + Level;
    }

    //4 дополнительных метода

    //1. Показать предмет
    public void ShowSubject()
    {
        Console.WriteLine("Предмет: " + Subject);
    }

    //2. Показать количество упражнений
    public void ShowExercises()
    {
        Console.WriteLine("Количество упражнений: " + ExercisesCount);
    }

    //3. Показать наличие ответов
    public void ShowAnswers()
    {
        if (HasAnswers)
        {
            Console.WriteLine("В учебнике есть ответы.");
        }
        else
        {
            Console.WriteLine("В учебнике нет ответов.");
        }
    }

    //4. Показать кафедру
    public void ShowDepartment()
    {
        Console.WriteLine("Кафедра: " + Department);
    }
}

//Сложное наследование
//AdvancedTextbook наследуется от Textbook,
//а Textbook уже наследуется от Book
class AdvancedTextbook : Textbook
{
    //4 дополнительных атрибута
    public string Difficulty { get; set; }
    public string Certification { get; set; }
    public int PracticalTasks { get; set; }
    public bool HasTests { get; set; }

    public AdvancedTextbook(string title, string author, int year,
                            int pages, string publisher,
                            string language, double rating,
                            string subject, string level,
                            int exercisesCount, bool hasAnswers,
                            string department,
                            string difficulty, string certification,
                            int practicalTasks, bool hasTests)
        : base(title, author, year, pages, publisher,
               language, rating, subject, level,
               exercisesCount, hasAnswers, department)
    {
        Difficulty = difficulty;
        Certification = certification;
        PracticalTasks = practicalTasks;
        HasTests = hasTests;
    }

    //Перекрытие метода Read
    public override void Read()
    {
        Console.WriteLine("Углублённый учебник " + Title +
                          " изучается по предмету: " + Subject);
    }

    //4 дополнительных метода

    //1. Показать сложность
    public void ShowDifficulty()
    {
        Console.WriteLine("Сложность: " + Difficulty);
    }

    //2. Показать сертификацию
    public void ShowCertification()
    {
        Console.WriteLine("Сертификация: " + Certification);
    }

    //3. Показать количество практических заданий
    public void ShowPracticalTasks()
    {
        Console.WriteLine("Практических заданий: " + PracticalTasks);
    }

    //4. Показать наличие тестов
    public void ShowTests()
    {
        if (HasTests)
        {
            Console.WriteLine("В учебнике есть тесты.");
        }
        else
        {
            Console.WriteLine("В учебнике нет тестов.");
        }
    }
}

//Производный класс Fiction
class Fiction : Book
{
    public string Genre { get; set; }

    //4 дополнительных атрибута
    public string Country { get; set; }
    public int Chapters { get; set; }
    public bool IsPopular { get; set; }
    public string MainCharacter { get; set; }

    public Fiction(string title, string author, int year,
                   int pages, string publisher,
                   string language, double rating,
                   string genre, string country,
                   int chapters, bool isPopular,
                   string mainCharacter)
        : base(title, author, year, pages, publisher, language, rating)
    {
        Genre = genre;
        Country = country;
        Chapters = chapters;
        IsPopular = isPopular;
        MainCharacter = mainCharacter;
    }

    //Перекрытие метода Borrow
    public override void Borrow()
    {
        Console.WriteLine("Художественная книга " + Title +
                          " выдана. Жанр: " + Genre);
    }

    //4 дополнительных метода

    //1. Показать жанр
    public void ShowGenre()
    {
        Console.WriteLine("Жанр: " + Genre);
    }

    //2. Показать страну
    public void ShowCountry()
    {
        Console.WriteLine("Страна: " + Country);
    }

    //3. Показать количество глав
    public void ShowChapters()
    {
        Console.WriteLine("Количество глав: " + Chapters);
    }

    //4. Показать популярность
    public void ShowPopularity()
    {
        if (IsPopular)
        {
            Console.WriteLine("Книга является популярной.");
        }
        else
        {
            Console.WriteLine("Книга не является популярной.");
        }
    }
}

//Производный класс ScientificLiterature
class ScientificLiterature : Book
{
    public string FieldOfScience { get; set; }

    //4 дополнительных атрибута
    public string Degree { get; set; }
    public int Citations { get; set; }
    public bool IsResearch { get; set; }
    public string Institution { get; set; }

    public ScientificLiterature(string title, string author,
                                int year, int pages,
                                string publisher,
                                string language, double rating,
                                string field, string degree,
                                int citations, bool isResearch,
                                string institution)
        : base(title, author, year, pages, publisher,
               language, rating)
    {
        FieldOfScience = field;
        Degree = degree;
        Citations = citations;
        IsResearch = isResearch;
        Institution = institution;
    }

    //Перекрытие метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Область науки: " + FieldOfScience;
    }

    //4 дополнительных метода

    //1. Показать область науки
    public void ShowField()
    {
        Console.WriteLine("Область науки: " + FieldOfScience);
    }

    //2. Показать учёную степень
    public void ShowDegree()
    {
        Console.WriteLine("Учёная степень автора: " + Degree);
    }

    //3. Показать количество цитирований
    public void ShowCitations()
    {
        Console.WriteLine("Количество цитирований: " + Citations);
    }

    //4. Показать научное учреждение
    public void ShowInstitution()
    {
        Console.WriteLine("Научное учреждение: " + Institution);
    }
}

//Множественное наследование через интерфейсы
//DigitalBook наследуется от Book и реализует два интерфейса
class DigitalBook : Book, IAudioBook, IDownloadable
{
    public string FileFormat { get; set; }

    //4 дополнительных атрибута
    public double FileSize { get; set; }
    public bool HasAudio { get; set; }
    public string DeviceType { get; set; }
    public bool IsDownloaded { get; set; }

    public DigitalBook(string title, string author,
                       int year, int pages,
                       string publisher,
                       string language, double rating,
                       string fileFormat, double fileSize,
                       bool hasAudio, string deviceType,
                       bool isDownloaded)
        : base(title, author, year, pages,
               publisher, language, rating)
    {
        FileFormat = fileFormat;
        FileSize = fileSize;
        HasAudio = hasAudio;
        DeviceType = deviceType;
        IsDownloaded = isDownloaded;
    }

    //Перекрытие метода GetInfo
    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Формат: " + FileFormat +
               ", Размер: " + FileSize + " МБ";
    }

    //Прослушивание книги
    public void Listen()
    {
        if (HasAudio)
        {
            Console.WriteLine("Аудиоверсия книги " +
                              Title + " запущена.");
        }
        else
        {
            Console.WriteLine("У книги нет аудиоверсии.");
        }
    }

    //Скачивание книги
    public void Download()
    {
        IsDownloaded = true;

        Console.WriteLine("Книга " + Title +
                          " скачана в формате " + FileFormat + ".");
    }

    //Явная реализация интерфейса IAudioBook
    void IAudioBook.Listen()
    {
        Listen();
    }

    //Явная реализация интерфейса IDownloadable
    void IDownloadable.Download()
    {
        Download();
    }

    //Показать размер файла
    public void ShowFileSize()
    {
        Console.WriteLine("Размер файла: " +
                          FileSize + " МБ");
    }

    //Показать устройство
    public void ShowDevice()
    {
        Console.WriteLine("Рекомендуемое устройство: " +
                          DeviceType);
    }
}

//Обобщённый класс для хранения книг
//T может быть Book или любым классом, который от него наследуется
class BookCollection<T> where T : Book
{
    private List<T> books = new List<T>();

    //Добавление книги в коллекцию
    public void Add(T book)
    {
        books.Add(book);
    }

    //Удаление книги из коллекции
    public void Remove(T book)
    {
        books.Remove(book);
    }

    //Вывод информации обо всех книгах
    public void DisplayBooks()
    {
        foreach (T book in books)
        {
            Console.WriteLine(book.GetInfo());
        }
    }

    //Поиск книг по автору
    public void FindByAuthor(string author)
    {
        bool found = false;

        foreach (T book in books)
        {
            if (book.Author == author)
            {
                Console.WriteLine(book.GetInfo());
                found = true;
            }
        }

        if (!found)
        {
            Console.WriteLine("Книги этого автора не найдены.");
        }
    }

    //Вывод количества книг в коллекции
    public void ShowCount()
    {
        Console.WriteLine("Количество книг в коллекции: " + books.Count);
    }
}

//Класс для управления книгами
//Зависимость передаётся через интерфейс IBookActions
class BookService
{
    private readonly IBookActions _book;

    //Конструктор принимает зависимость
    public BookService(IBookActions book)
    {
        _book = book;
    }

    //Выполнение действий с книгой
    public void UseBook()
    {
        _book.ShowShortInfo();
        _book.ReadBook();
    }
}

//Создание объектов

Textbook textbook = new Textbook(
    "Программирование на C#",
    "Спирин И.С.",
    2024,
    450,
    "Питер",
    "Русский",
    9.5,
    "Информатика",
    "Для студентов",
    120,
    true,
    "Кафедра информатики"
);

AdvancedTextbook advancedTextbook = new AdvancedTextbook(
    "Продвинутое программирование",
    "Иванов А.А.",
    2025,
    600,
    "БХВ-Петербург",
    "Русский",
    9.8,
    "Программирование",
    "Продвинутый",
    200,
    true,
    "Кафедра программирования",
    "Высокая",
    "Профессиональная",
    80,
    true
);

Fiction fiction = new Fiction(
    "Война и мир",
    "Лев Толстой",
    1869,
    1300,
    "Эксмо",
    "Русский",
    10.0,
    "Роман",
    "Россия",
    15,
    true,
    "Пьер Безухов"
);

ScientificLiterature science = new ScientificLiterature(
    "Основы физики",
    "Степанов А.М.",
    2023,
    500,
    "Наука",
    "Русский",
    9.0,
    "Физика",
    "Доктор наук",
    85,
    true,
    "МГУ"
);

DigitalBook digitalBook = new DigitalBook(
    "Изучаем C#",
    "Петров А.А.",
    2025,
    350,
    "БХВ-Петербург",
    "Русский",
    9.7,
    "PDF",
    15.5,
    true,
    "Компьютер",
    false
);

//Проверка методов

Console.WriteLine("Информация об учебнике:");
Console.WriteLine(textbook.GetInfo());
textbook.Read();
textbook.Borrow();

textbook.ShowPages();
textbook.ShowPublisher();
textbook.ShowLanguage();
textbook.ShowRating();

textbook.ShowSubject();
textbook.ShowExercises();
textbook.ShowAnswers();
textbook.ShowDepartment();

Console.WriteLine();

Console.WriteLine("Информация об углублённом учебнике:");
Console.WriteLine(advancedTextbook.GetInfo());
advancedTextbook.Read();

advancedTextbook.ShowDifficulty();
advancedTextbook.ShowCertification();
advancedTextbook.ShowPracticalTasks();
advancedTextbook.ShowTests();

Console.WriteLine();

Console.WriteLine("Информация о художественной книге:");
Console.WriteLine(fiction.GetInfo());
fiction.Read();
fiction.Borrow();

fiction.ShowGenre();
fiction.ShowCountry();
fiction.ShowChapters();
fiction.ShowPopularity();

Console.WriteLine();

Console.WriteLine("Информация о научной литературе:");
Console.WriteLine(science.GetInfo());
science.Read();
science.Borrow();

science.ShowField();
science.ShowDegree();
science.ShowCitations();
science.ShowInstitution();

Console.WriteLine();

Console.WriteLine("Информация об электронной книге:");
Console.WriteLine(digitalBook.GetInfo());
digitalBook.Read();
digitalBook.Borrow();

digitalBook.Listen();
digitalBook.Download();
digitalBook.ShowFileSize();
digitalBook.ShowDevice();

//Проверка явной реализации интерфейсов

Console.WriteLine();
Console.WriteLine("Явная реализация интерфейсов:");

//Создание ссылок на интерфейсы
IAudioBook audioBook = digitalBook;
IDownloadable downloadableBook = digitalBook;

//Вызов методов через интерфейсы
audioBook.Listen();
downloadableBook.Download();

//Взаимодействие объектов между собой

Console.WriteLine();
Console.WriteLine("Взаимодействие объектов:");

//Сравнение годов издания
textbook.CompareYear(fiction);
science.CompareYear(textbook);
digitalBook.CompareYear(science);

//Рекомендация книг
textbook.Recommend(science);
fiction.Recommend(textbook);

//Проверка перегрузки методов

Console.WriteLine();
Console.WriteLine("Перегрузка методов:");

//Чтение без параметров
textbook.Read();

//Чтение определенного количества страниц
textbook.Read(20);

//Чтение в указанном формате
textbook.Read("PDF");

//Сравнение с указанным годом
textbook.CompareYear(2020);

//Проверка полиморфизма

Console.WriteLine();
Console.WriteLine("Полиморфизм:");

//Объекты разных классов обрабатываются через базовый класс Book
List<Book> bookList = new List<Book>
{
    textbook,
    advancedTextbook,
    fiction,
    science,
    digitalBook
};

foreach (Book book in bookList)
{
    Console.WriteLine(book.GetInfo());
    book.Read();
    Console.WriteLine();
}

//Проверка обобщённого класса

Console.WriteLine();
Console.WriteLine("Обобщённая коллекция книг:");

BookCollection<Book> collection = new BookCollection<Book>();

//Добавление книг в коллекцию
collection.Add(textbook);
collection.Add(advancedTextbook);
collection.Add(fiction);
collection.Add(science);
collection.Add(digitalBook);

//Вывод информации обо всех книгах
collection.DisplayBooks();

//Вывод количества книг
collection.ShowCount();

//Поиск книг по автору
Console.WriteLine();
Console.WriteLine("Поиск книг автора Лев Толстой:");

collection.FindByAuthor("Лев Толстой");

//Удаление книги из коллекции
Console.WriteLine();
Console.WriteLine("Удаление книги из коллекции:");

collection.Remove(fiction);

//Повторный вывод количества книг
collection.ShowCount();

//Проверка явной реализации интерфейса IBookActions

Console.WriteLine();
Console.WriteLine("Явная реализация IBookActions:");

//Создание ссылки на интерфейс
IBookActions bookActions = textbook;

//Вызов явных методов через интерфейс
bookActions.ShowShortInfo();
bookActions.ReadBook();

//Проверка управления зависимостями

Console.WriteLine();
Console.WriteLine("Управление зависимостями:");

//Передача разных книг в сервис через конструктор
BookService textbookService = new BookService(textbook);
BookService fictionService = new BookService(fiction);
BookService scienceService = new BookService(science);
BookService digitalBookService = new BookService(digitalBook);

//Использование сервисов
Console.WriteLine();
Console.WriteLine("Работа с учебником:");
textbookService.UseBook();

Console.WriteLine();
Console.WriteLine("Работа с художественной книгой:");
fictionService.UseBook();

Console.WriteLine();
Console.WriteLine("Работа с научной литературой:");
scienceService.UseBook();

Console.WriteLine();
Console.WriteLine("Работа с электронной книгой:");
digitalBookService.UseBook();
```

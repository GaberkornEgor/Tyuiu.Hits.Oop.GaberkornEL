<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();
        displayBike();
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
//Создаем делегат для вывода информации
public delegate void DisplayDelegate();

//Базовый класс транспорта
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    //Выводим информацию о транспорте
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }

    //Изменяем модель транспорта
    public void UpdateModel(string newModel)
    {
        Model = newModel;
        TransportUpdated?.Invoke($"Модель обновлена на: {newModel}");
    }
}

//Класс автомобиля
public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

//Класс велосипеда
public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Наличие багажника: {(HasCarrier ? "Есть" : "Нет")}");
    }
}

//Создаем объекты транспорта
Car car = new Car("Model S", "Tesla", 4);
Bike bike = new Bike("Speedster", "Giant", true);

Console.WriteLine("=== Делегаты ===");

//Присваиваем делегату методы DisplayInfo
DisplayDelegate displayCar = car.DisplayInfo;
DisplayDelegate displayBike = bike.DisplayInfo;

//Вызываем методы через делегаты
displayCar();
displayBike();

Console.WriteLine();

//Создаем список транспорта
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
    new Car("Civic", "Honda", 4)
};

Console.WriteLine("=== Лямбда-выражения ===");

//Находим автомобили производителя Tesla
var teslaCars = transports.OfType<Car>()
                          .Where(c => c.Manufacturer == "Tesla");

//Выводим найденные автомобили
foreach (var item in teslaCars)
{
    item.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine("=== События ===");

//Создаем объект транспорта
Transport transport = new Transport("Старая модель", "Неизвестный производитель");

//Подписываемся на событие
transport.TransportUpdated += message => Console.WriteLine(message);

//Изменяем модель
transport.UpdateModel("Новая модель");

Console.WriteLine();

//Выводим обновленную информацию
transport.DisplayInfo();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
//Делегат для вывода информации о животных
public delegate void DisplayDelegate();

//Базовый класс животного
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    //Делегат для события
    public delegate void AnimalHandler(string message);

    //Событие изменения информации
    public event AnimalHandler AnimalUpdated;

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    //Вывод информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Возраст: {Age}");
        Console.WriteLine($"Вес: {Weight} кг");
    }

    //Обновление имени животного
    public void UpdateName(string newName)
    {
        Name = newName;

        //Вызываем событие
        AnimalUpdated?.Invoke($"Имя животного изменено на: {newName}");
    }

    //Обновление возраста
    public void UpdateAge(int newAge)
    {
        Age = newAge;

        //Вызываем событие
        AnimalUpdated?.Invoke($"Возраст животного изменен на: {newAge}");
    }
}

//Класс собаки
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
        Console.WriteLine("Вид: Собака");
    }

    public void Bark()
    {
        Console.WriteLine($"{Name} говорит: Гав!");
    }
}

//Класс кошки
public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, double weight, string color)
        : base(name, age, weight)
    {
        Color = color;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}");
        Console.WriteLine("Вид: Кошка");
    }

    public void Meow()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }
}

//Класс птицы
public class Bird : Animal
{
    public double Wingspan { get; set; }

    public Bird(string name, int age, double weight, double wingspan)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
        Console.WriteLine("Вид: Птица");
    }

    public void Fly()
    {
        Console.WriteLine($"{Name} летает.");
    }
}

//Класс рыбы
public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, int age, double weight, string waterType)
        : base(name, age, weight)
    {
        WaterType = waterType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
        Console.WriteLine("Вид: Рыба");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает.");
    }
}

//Создаем объекты животных
Dog dog = new Dog("Шарик", 3, 12.5, "Овчарка");
Cat cat = new Cat("Мурка", 2, 4.2, "Рыжий");
Bird bird = new Bird("Кеша", 1, 0.5, 25);
Fish fish = new Fish("Немо", 1, 0.2, "Морская");

//Делегаты

Console.WriteLine("=== Делегаты ===");

//Присваиваем делегатам методы
DisplayDelegate displayDog = dog.DisplayInfo;
DisplayDelegate displayCat = cat.DisplayInfo;
DisplayDelegate displayBird = bird.DisplayInfo;
DisplayDelegate displayFish = fish.DisplayInfo;

//Вызываем методы через делегаты
displayDog();
Console.WriteLine();

displayCat();
Console.WriteLine();

displayBird();
Console.WriteLine();

displayFish();
Console.WriteLine();

//Лямбда-выражения

Console.WriteLine("=== Лямбда-выражения ===");

//Создаем список животных
List<Animal> animals = new List<Animal>
{
    dog,
    cat,
    bird,
    fish
};

//Находим животных старше одного года
var adultAnimals = animals.Where(a => a.Age > 1);

Console.WriteLine("Животные старше одного года:");

foreach (var animal in adultAnimals)
{
    Console.WriteLine(animal.Name);
}

//Находим собак
var dogs = animals.OfType<Dog>();

Console.WriteLine();
Console.WriteLine("Список собак:");

foreach (var animal in dogs)
{
    Console.WriteLine(animal.Name);
}

//Находим животных весом меньше 5 кг
var smallAnimals = animals.Where(a => a.Weight < 5);

Console.WriteLine();
Console.WriteLine("Животные весом меньше 5 кг:");

foreach (var animal in smallAnimals)
{
    Console.WriteLine(animal.Name);
}

//События

Console.WriteLine();
Console.WriteLine("=== События ===");

//Подписываемся на событие собаки
dog.AnimalUpdated += message =>
{
    Console.WriteLine(message);
};

//Подписываемся на событие кошки
cat.AnimalUpdated += message =>
{
    Console.WriteLine(message);
};

//Изменяем данные животных
dog.UpdateName("Рекс");
cat.UpdateAge(3);

//Выводим обновленную информацию
Console.WriteLine();
Console.WriteLine("=== Обновленная информация ===");

dog.DisplayInfo();
Console.WriteLine();

cat.DisplayInfo();
````

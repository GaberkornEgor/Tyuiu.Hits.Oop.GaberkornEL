Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;


//Базовый класс Book
class Book
{
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }

    //4 дополнительных атрибута
    public int Pages { get; set; }
    public string Publisher { get; set; }
    public string Language { get; set; }
    public double Rating { get; set; }

    //Конструктор
    public Book(string title, string author, int year,
                int pages, string publisher,
                string language, double rating)
    {
        Title = title;
        Author = author;
        YearOfPublication = year;
        Pages = pages;
        Publisher = publisher;
        Language = language;
        Rating = rating;
    }

    public virtual string GetInfo()
    {
        return "Название: " + Title +
               ", Автор: " + Author +
               ", Год издания: " + YearOfPublication;
    }

    public virtual void Read()
    {
        Console.WriteLine("Книга " + Title + " читается.");
    }

    public virtual void Borrow()
    {
        Console.WriteLine("Книга " + Title + " выдана на чтение.");
    }


    //4 дополнительных метода

    //1. Вывод количества страниц
    public void ShowPages()
    {
        Console.WriteLine("Количество страниц: " + Pages);
    }

    //2. Вывод издательства
    public void ShowPublisher()
    {
        Console.WriteLine("Издательство: " + Publisher);
    }

    //3. Вывод языка
    public void ShowLanguage()
    {
        Console.WriteLine("Язык книги: " + Language);
    }

    //4. Вывод рейтинга
    public void ShowRating()
    {
        Console.WriteLine("Рейтинг книги: " + Rating);
    }


    //Взаимодействие объектов:
    //один объект Book сравнивает себя с другим объектом Book
    public void CompareYear(Book book)
    {
        if (YearOfPublication > book.YearOfPublication)
        {
            Console.WriteLine(Title + " новее, чем " + book.Title);
        }
        else if (YearOfPublication < book.YearOfPublication)
        {
            Console.WriteLine(Title + " старше, чем " + book.Title);
        }
        else
        {
            Console.WriteLine(Title + " и " + book.Title +
                              " изданы в одном году.");
        }
    }

    //Взаимодействие объектов:
    //один объект Book взаимодействует с другим объектом Book
    public void Recommend(Book book)
    {
        Console.WriteLine("Книга " + Title +
                          " рекомендуется вместе с книгой " + book.Title);
    }
}


//Простое наследование
//Textbook наследуется от Book
class Textbook : Book
{
    public string Subject { get; set; }

    //4 дополнительных атрибута
    public string Level { get; set; }
    public int ExercisesCount { get; set; }
    public bool HasAnswers { get; set; }
    public string Department { get; set; }

    public Textbook(string title, string author, int year,
                    int pages, string publisher,
                    string language, double rating,
                    string subject, string level,
                    int exercisesCount, bool hasAnswers,
                    string department)
        : base(title, author, year, pages, publisher, language, rating)
    {
        Subject = subject;
        Level = level;
        ExercisesCount = exercisesCount;
        HasAnswers = hasAnswers;
        Department = department;
    }

    public override void Read()
    {
        Console.WriteLine("Учебник " + Title +
                          " читается по предмету: " + Subject);
    }

    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Предмет: " + Subject +
               ", Уровень: " + Level;
    }


    //4 дополнительных метода

    //1. Показать предмет
    public void ShowSubject()
    {
        Console.WriteLine("Предмет: " + Subject);
    }

    //2. Показать количество упражнений
    public void ShowExercises()
    {
        Console.WriteLine("Количество упражнений: " + ExercisesCount);
    }

    //3. Показать наличие ответов
    public void ShowAnswers()
    {
        if (HasAnswers)
        {
            Console.WriteLine("В учебнике есть ответы.");
        }
        else
        {
            Console.WriteLine("В учебнике нет ответов.");
        }
    }

    //4. Показать кафедру
    public void ShowDepartment()
    {
        Console.WriteLine("Кафедра: " + Department);
    }
}


//Сложное наследование
//AdvancedTextbook наследуется от Textbook,
//а Textbook уже наследуется от Book
class AdvancedTextbook : Textbook
{
    //4 дополнительных атрибута
    public string Difficulty { get; set; }
    public string Certification { get; set; }
    public int PracticalTasks { get; set; }
    public bool HasTests { get; set; }

    public AdvancedTextbook(string title, string author, int year,
                            int pages, string publisher,
                            string language, double rating,
                            string subject, string level,
                            int exercisesCount, bool hasAnswers,
                            string department,
                            string difficulty, string certification,
                            int practicalTasks, bool hasTests)
        : base(title, author, year, pages, publisher,
               language, rating, subject, level,
               exercisesCount, hasAnswers, department)
    {
        Difficulty = difficulty;
        Certification = certification;
        PracticalTasks = practicalTasks;
        HasTests = hasTests;
    }

    public override void Read()
    {
        Console.WriteLine("Углублённый учебник " + Title +
                          " изучается по предмету: " + Subject);
    }


    //4 дополнительных метода

    //1. Показать сложность
    public void ShowDifficulty()
    {
        Console.WriteLine("Сложность: " + Difficulty);
    }

    //2. Показать сертификацию
    public void ShowCertification()
    {
        Console.WriteLine("Сертификация: " + Certification);
    }

    //3. Показать количество практических заданий
    public void ShowPracticalTasks()
    {
        Console.WriteLine("Практических заданий: " + PracticalTasks);
    }

    //4. Показать наличие тестов
    public void ShowTests()
    {
        if (HasTests)
        {
            Console.WriteLine("В учебнике есть тесты.");
        }
        else
        {
            Console.WriteLine("В учебнике нет тестов.");
        }
    }
}


//Производный класс Fiction
class Fiction : Book
{
    public string Genre { get; set; }

    //4 дополнительных атрибута
    public string Country { get; set; }
    public int Chapters { get; set; }
    public bool IsPopular { get; set; }
    public string MainCharacter { get; set; }

    public Fiction(string title, string author, int year,
                   int pages, string publisher,
                   string language, double rating,
                   string genre, string country,
                   int chapters, bool isPopular,
                   string mainCharacter)
        : base(title, author, year, pages, publisher, language, rating)
    {
        Genre = genre;
        Country = country;
        Chapters = chapters;
        IsPopular = isPopular;
        MainCharacter = mainCharacter;
    }

    public override void Borrow()
    {
        Console.WriteLine("Художественная книга " + Title +
                          " выдана. Жанр: " + Genre);
    }


    //4 дополнительных метода

    //1. Показать жанр
    public void ShowGenre()
    {
        Console.WriteLine("Жанр: " + Genre);
    }

    //2. Показать страну
    public void ShowCountry()
    {
        Console.WriteLine("Страна: " + Country);
    }

    //3. Показать количество глав
    public void ShowChapters()
    {
        Console.WriteLine("Количество глав: " + Chapters);
    }

    //4. Показать популярность
    public void ShowPopularity()
    {
        if (IsPopular)
        {
            Console.WriteLine("Книга является популярной.");
        }
        else
        {
            Console.WriteLine("Книга не является популярной.");
        }
    }
}


//Производный класс ScientificLiterature
class ScientificLiterature : Book
{
    public string FieldOfScience { get; set; }

    //4 дополнительных атрибута
    public string Degree { get; set; }
    public int Citations { get; set; }
    public bool IsResearch { get; set; }
    public string Institution { get; set; }

    public ScientificLiterature(string title, string author,
                                int year, int pages,
                                string publisher,
                                string language, double rating,
                                string field, string degree,
                                int citations, bool isResearch,
                                string institution)
        : base(title, author, year, pages, publisher,
               language, rating)
    {
        FieldOfScience = field;
        Degree = degree;
        Citations = citations;
        IsResearch = isResearch;
        Institution = institution;
    }

    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Область науки: " + FieldOfScience;
    }


    //4 дополнительных метода

    //1. Показать область науки
    public void ShowField()
    {
        Console.WriteLine("Область науки: " + FieldOfScience);
    }

    //2. Показать учёную степень
    public void ShowDegree()
    {
        Console.WriteLine("Учёная степень автора: " + Degree);
    }

    //3. Показать количество цитирований
    public void ShowCitations()
    {
        Console.WriteLine("Количество цитирований: " + Citations);
    }

    //4. Показать научное учреждение
    public void ShowInstitution()
    {
        Console.WriteLine("Научное учреждение: " + Institution);
    }
}


//Множественное наследование
//В C# нельзя наследоваться сразу от нескольких классов,
//поэтому для множественного наследования используются интерфейсы.

interface IAudioBook
{
    void Listen();
}

interface IDownloadable
{
    void Download();
}


//DigitalBook наследуется от Book
//и реализует два интерфейса
class DigitalBook : Book, IAudioBook, IDownloadable
{
    public string FileFormat { get; set; }

    //4 дополнительных атрибута
    public double FileSize { get; set; }
    public bool HasAudio { get; set; }
    public string DeviceType { get; set; }
    public bool IsDownloaded { get; set; }

    public DigitalBook(string title, string author,
                       int year, int pages,
                       string publisher,
                       string language, double rating,
                       string fileFormat, double fileSize,
                       bool hasAudio, string deviceType,
                       bool isDownloaded)
        : base(title, author, year, pages,
               publisher, language, rating)
    {
        FileFormat = fileFormat;
        FileSize = fileSize;
        HasAudio = hasAudio;
        DeviceType = deviceType;
        IsDownloaded = isDownloaded;
    }

    public override string GetInfo()
    {
        return base.GetInfo() +
               ", Формат: " + FileFormat +
               ", Размер: " + FileSize + " МБ";
    }


    //4 дополнительных метода

    //1. Прослушивание книги
    public void Listen()
    {
        if (HasAudio)
        {
            Console.WriteLine("Аудиоверсия книги " +
                              Title + " запущена.");
        }
        else
        {
            Console.WriteLine("У книги нет аудиоверсии.");
        }
    }

    //2. Скачивание книги
    public void Download()
    {
        IsDownloaded = true;

        Console.WriteLine("Книга " + Title +
                          " скачана в формате " + FileFormat + ".");
    }

    //3. Показать размер файла
    public void ShowFileSize()
    {
        Console.WriteLine("Размер файла: " +
                          FileSize + " МБ");
    }

    //4. Показать устройство
    public void ShowDevice()
    {
        Console.WriteLine("Рекомендуемое устройство: " +
                          DeviceType);
    }
}


//Создание объектов

Textbook textbook = new Textbook(
    "Программирование на C#",
    "Спирин И.С.",
    2024,
    450,
    "Питер",
    "Русский",
    9.5,
    "Информатика",
    "Для студентов",
    120,
    true,
    "Кафедра информатики"
);


AdvancedTextbook advancedTextbook = new AdvancedTextbook(
    "Продвинутое программирование",
    "Иванов А.А.",
    2025,
    600,
    "БХВ-Петербург",
    "Русский",
    9.8,
    "Программирование",
    "Продвинутый",
    200,
    true,
    "Кафедра программирования",
    "Высокая",
    "Профессиональная",
    80,
    true
);


Fiction fiction = new Fiction(
    "Война и мир",
    "Лев Толстой",
    1869,
    1300,
    "Эксмо",
    "Русский",
    10.0,
    "Роман",
    "Россия",
    15,
    true,
    "Пьер Безухов"
);


ScientificLiterature science = new ScientificLiterature(
    "Основы физики",
    "Степанов А.М.",
    2023,
    500,
    "Наука",
    "Русский",
    9.0,
    "Физика",
    "Доктор наук",
    85,
    true,
    "МГУ"
);


DigitalBook digitalBook = new DigitalBook(
    "Изучаем C#",
    "Петров А.А.",
    2025,
    350,
    "БХВ-Петербург",
    "Русский",
    9.7,
    "PDF",
    15.5,
    true,
    "Компьютер",
    false
);


//Проверка методов

Console.WriteLine(textbook.GetInfo());
textbook.Read();
textbook.Borrow();

textbook.ShowPages();
textbook.ShowPublisher();
textbook.ShowLanguage();
textbook.ShowRating();

textbook.ShowSubject();
textbook.ShowExercises();
textbook.ShowAnswers();
textbook.ShowDepartment();


Console.WriteLine();

Console.WriteLine(advancedTextbook.GetInfo());
advancedTextbook.Read();

advancedTextbook.ShowDifficulty();
advancedTextbook.ShowCertification();
advancedTextbook.ShowPracticalTasks();
advancedTextbook.ShowTests();


Console.WriteLine();

Console.WriteLine(fiction.GetInfo());
fiction.Read();
fiction.Borrow();

fiction.ShowGenre();
fiction.ShowCountry();
fiction.ShowChapters();
fiction.ShowPopularity();


Console.WriteLine();

Console.WriteLine(science.GetInfo());
science.Read();
science.Borrow();

science.ShowField();
science.ShowDegree();
science.ShowCitations();
science.ShowInstitution();


Console.WriteLine();

Console.WriteLine(digitalBook.GetInfo());
digitalBook.Read();
digitalBook.Borrow();

digitalBook.Listen();
digitalBook.Download();
digitalBook.ShowFileSize();
digitalBook.ShowDevice();


//Взаимодействие объектов между собой

Console.WriteLine();
Console.WriteLine("Взаимодействие объектов:");


//Объект textbook взаимодействует с объектом fiction
//и сравнивает их годы издания
textbook.CompareYear(fiction);


//Объект science взаимодействует с объектом textbook
//и сравнивает их годы издания
science.CompareYear(textbook);


//Объект textbook взаимодействует с объектом science
//и рекомендует его вместе с собой
textbook.Recommend(science);


//Объект fiction взаимодействует с объектом textbook
//и рекомендует его вместе с собой
fiction.Recommend(textbook);


//Объект digitalBook взаимодействует с объектом science
digitalBook.CompareYear(science);
```
